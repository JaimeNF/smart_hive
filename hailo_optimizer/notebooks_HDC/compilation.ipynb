# Compilation code

```python
import os
os.environ["CUDA_VISIBLE_DEVICES"] = "1"
from hailo_sdk_client import ClientRunner


# Model  ID
MODEL_NAME = "yolov11n"
EXPERIMENT_TAG = "max_FPS"

# Local paths
BASE_DIR = ".."
PATHS = {
    # Input: quantize model
    "quantized_har": f"{BASE_DIR}/models/intermediate/{MODEL_NAME}_{EXPERIMENT_TAG}_quantized.har",
    # Output: binaries
    "compiled_hef":  f"{BASE_DIR}/models/binaries/{MODEL_NAME}_{EXPERIMENT_TAG}.hef",
}

os.makedirs(os.path.dirname(PATHS['compiled_hef']), exist_ok=True)
```

```python
def compile_model(har_path, hef_path):
    print(f"🔨 Compilando modelo desde: {har_path}")
    runner = ClientRunner(har=har_path)

    hef = runner.compile()

    with open(hef_path, 'wb') as f:
        f.write(hef)

    print(f"✅ Compilación exitosa. Guardado en: {hef_path}")

# Ejecutar compilación
if os.path.exists(PATHS['quantized_har']):
    compile_model(PATHS['quantized_har'], PATHS['compiled_hef'])
else:
    print(f"❌ Error: No se encuentra el modelo cuantizado {PATHS['quantized_har']}")
```
